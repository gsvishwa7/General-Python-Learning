## DataFrames

In the interest of time, I am going to skip over Series since they are *analogous* to DataFrames with one column. Yes, I acknowledge that this is not strictly true. 

### Creating a Pandas DataFrame


First, we create and print our first dataframe.

```python
import pandas as pd

data = {
    "Name":["Girish", "Priyanka", "Joergen", "Emily", "Yiling"], 
    "Age": [27, 28, 30, 27, 25]
}

my_first_df = pd.DataFrame(data)

print(my_first_df)
```

### Adding an index

Next, we manually set an index for each of our entries and show how we can access specific entries using the custom index and the default integer index.

```python
my_indexed_df = pd.DataFrame(data, index=["Crewmate 1", "Crewmate 2", "Crewmate 3", "Crewmate 4",  "Crewmate 5"])
print(my_indexed_df, "\n")

# Comment one out as needed
my_indexed_df.loc["Crewmate 4"] # The data type is "object"
my_indexed_df.iloc[2] # iloc is "integer location" essentially
```

Pandas indeed indexes your entries using positive integers by default. So, shouldn't  `my_first_df` also be accessible using integer labels?

```python
my_first_df.loc[2] # Indeed, this works! No need to use iloc since no custom index was given
```

### Add a new columm
Next, we add a new column to `my_indexed_df`.

```python
my_indexed_df["Job"] = ["Unemployed", "Surgeon", "Physicist", "Software Engineer", "PhD Student"] # Creates a new column called "Job" 
print(my_indexed_df, "\n")
```

### Add new row(s)
We now create a new row and add it to the end of the dataframe. __NOTE:__ Running this cell mulitple times creates duplicate rows.

```python
new_row = pd.DataFrame([{"Name": "Áine", "Age": 28, "Job": "Neuroscientist"}]) # Creates a new dataframe consisting of only one row
print(new_row, "\n") # Oh, we forgot our index, so the new entry is just labelled by 0

new_row = pd.DataFrame([{"Name": "Áine", "Age": 28, "Job": "Neuroscientist"}], index=["Crewmate 6"]) # Same new row, just added an index
print(new_row, "\n")

my_indexed_df = pd.concat([my_indexed_df, new_row])
print(my_indexed_df, "\n")

# my_indexed_df.drop("Crewmate 6", inplace=True) # Removes all instances of "Crewmate 6", so if you duplicated this entry, all copies get removed
# print(my_indexed_df)
```

You may have noticed that the way we added five rows was much more efficient than adding the one new row. This is because we used a list of strings and integers for each key in the dictionary when creating the first five rows. We may do the same to add multiple rows at once.

```python
new_rows = pd.DataFrame({"Name": ["Antoni", "Anjali"], "Age": [28, 26], "Job": ["Statistician", "Junior Doctor (FY2)"]})
my_indexed_df = pd.concat([my_indexed_df, new_rows])
print(my_indexed_df, "\n") # Oops, forgot the index again

my_indexed_df.drop([0,1], inplace=True)

new_rows = pd.DataFrame({"Name": ["Antoni", "Anjali"], "Age": [28, 26], "Job": ["Statistician", "Junior Doctor (FY2)"]}, index=["Crewmate 7", "Crewmate 8"])
my_indexed_df = pd.concat([my_indexed_df, new_rows])
print(my_indexed_df, "\n")

# my_indexed_df.drop(["Crewmate 7", "Crewmate 8"], inplace= True)   # Again, in case you need it :)
```

## Importing from a CSV file

```python
my_import = pd.read_csv("C:/Users/priya/Downloads/Gen1_Pokemon.csv")

print(my_import)
```

### Selecting by columns

```python
print(my_import[["Name", "Type 1", "Type 2"]]) # Selecting by columns
```

### Selecting by rows

```python
my_indexed_import = pd.read_csv("C:/Users/priya/Downloads/Gen1_Pokemon.csv", index_col="Name") # Setting one of the columns of the DataFrame to be the indexing column itself

# This lets us access specific entries (i.e., rows) using the Pokemon name as index!
print(my_indexed_import.loc["Dragonite"])
```

```python
print(my_import.loc[[i for i in range(10)]]) # Selecting first 10 rows using integer location and list comprehension
```

```python
print(my_import.loc[[i for i in range(15, 20)], ["Name", "Attack", "Defense"]]) # We can also choose to show specific columns of the specific rows we would like to see

print(my_indexed_import.loc["Charizard": "Blastoise", ["Type 1", "Type 2", "HP"]]) # Slicing notation is compatible with custom-indexed rows too! NOTE the weird square brackets though...
```
