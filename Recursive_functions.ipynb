## Problem 1: Write a Python program to sum recursion lists using recursion.

Test Data: `[1, 2, [3,4], [5,6]]`

Expected Result: `21`

```python
def recursive_list_sum(my_list):
    total = 0
    for element in my_list:
        if type(element) == list:
            total += recursive_list_sum(element)
        else:
            total += element
    return total

print(recursive_list_sum([1, 2, [3,4], [5,6],[7,[10,2,12]]]))    
```

## Problem 2: Write a Python program to get the factorial of a non-negative integer using recursion

```python
def my_factorial(n):
    if n < 0 or type(n) != int:
        raise ValueError('n must be a positive integer')
    if n > 1:
        return n * my_factorial(n-1)
    elif n == 1:
        return 1

print(my_factorial(7))
```

## Problem 3: Write a Python program to solve the Fibonacci sequence using recursion. 

-  First, just write a function that spits out the nth term of the Fibonacci sequence. For example:

Input: `7`, Expected Result: `13`

-  Then, try to write a function that gives you the list of the first n terms fo the Fibonacci sequence. That is:

Input: `7`, Expected Result: `[1, 1, 2, 3, 5, 8, 13]`

```python
def fibonacci(n):
    if n == 1 or n == 2:
        return 1
    else:
        return fibonacci(n-1) + fibonacci(n-2)

def fibonacci_list(n):
    # output = []
    if n == 0:
        return []
    elif n == 1:
        return [1]
    elif n == 2:
        return [1,1]
    else:
        seq = fibonacci_list(n-1)
        seq.append(seq[-1] + seq[-2])
        return seq
    
# A more efficient version of fibonacci(n):
def fib_efficient(n, a=1, b=1):
    if n < 0 or type(n) != int:
        raise ValueError('n must be a positive integer')
    if n == 1:
        return a
    if n == 2:
        return b
    else:
        # One recursive call passing the next* two values forward. The reassignment is essentially the storage of the previous* 
        # (mathematically speaking, but indeed these are the next* two terms in the sequence when viewed algorithmically, 
        # since we are in some sense starting from n and working our way down) two terms of the sequence.
        return fib_efficient(n - 1, b, a + b)

### Test cases ###

# print(fibonacci(7))
# print(fibonacci_list(3)) # -- Note that n=3 is a good test case to take apart the recursive algorithm and understand what gets sent where.
# print (fib_efficient(3)) # -- Again, n=3 is a good test case to try and understand the algorithm well
```
